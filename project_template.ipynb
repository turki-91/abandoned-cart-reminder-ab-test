### التعريفات التشغيلية المقترحة (افتراضات)

**السلة المتروكة ووقت البداية:** نراقب سلة محفوظة وغير فارغة مرتبطة بحساب. يبدأ عداد عدم النشاط من آخر حدث مسجل يخصها: إضافة أو حذف منتج، تغيير كمية، عرض السلة أو خطوة دفع. إذا مضت 24 ساعة متصلة دون أي من هذه الأحداث، وبقيت السلة غير فارغة ولم يسجل الحساب طلبًا مدفوعًا منذ ذلك الحدث، تعد السلة متروكة. أي نشاط جديد يعيد العداد قبل الأهلية. عند بلوغ 24 ساعة نفحص الشروط الآتية على الخادم، ثم نوزع المستخدم فورًا؛ لحظة التوزيع هي `t0` في المجموعتين. إذا تزامن دفع مع التوزيع نستخدم ترتيب أحداث الخادم ونفحص الدفع أولًا.

**الأهلية قبل التوزيع فقط:** حساب مسجل بمعرف موحد عبر الأجهزة، بريد متحقق منه، وموافقة سارية على تلقي هذا النوع من الرسائل؛ السلة تحقق التعريف السابق؛ المستخدم ضمن السوق التجريبي الافتراضي (السعودية)؛ ليس حساب اختبار أو موظفًا أو بوتًا معروفًا؛ لا يوجد حظر بريدي أو ارتداد معروف؛ لم يتلق تذكير سلة خلال الأيام السبعة السابقة، ولا يشارك في حملة تذكير أو حافز متعارضة، ولم يسبق تسجيله في هذه التجربة. تستخدم جميع هذه الفحوص معلومات متاحة قبل `t0` وتطبق بالطريقة نفسها على المرشحين للمجموعتين. لا نشترط فتح رسالة أو نقرة أو أي سلوك لاحق للأهلية.

**A — Control:** لا نرسل تذكير سلة. تبقى خدمة المتجر ورسائل المعاملات الضرورية، مثل تأكيد الطلب، كما هي.

**B — Treatment:** محاولة إرسال بريد تذكير واحد فور التوزيع عند `t0` (مهلة تنفيذ تشغيلية حتى 5 دقائق)، بعنوان ثابت «سلتك محفوظة»، ونص ثابت «ما زالت منتجاتك في السلة. يمكنك العودة لإتمام الطلب»، ورابط شخصي يعيد المستخدم إلى سلته. لغة الرسالة العربية والقالب والتوقيت موحدة، مع رابط إلغاء الاشتراك المعتاد. **لا خصم ولا كوبون ولا توصيل مجاني إضافي ولا تغيير في الأسعار أو صفحة الدفع.** نوقف أي تذكير سلة آخر للمجموعتين طوال المتابعة. لا نعيد المحاولة إذا فشل الإرسال. نحترم سحب الموافقة أو الشراء المسجل قبل تنفيذ الرسالة ونلغي الإرسال عندهما، مع إبقاء التخصيص والسجل في التحليل.

**تخصيص ثابت:** بعد فحص الأهلية نستخدم منهج تمرين 1.2: `SHA-256(experiment_id:user_id)` مع معرف ثابت للتجربة `cart_reminder_v1` وحد ثابت 0.50 لمجموعة B، ونحفظ التخصيص في سجل مركزي. لا نعيد القرعة عند زيارة جديدة أو جهاز آخر، ولا نغير النسبة أثناء التجربة. لا يدخل المستخدم مجددًا بسبب سلة أخرى ولا يحصل على تذكير ثان. 50/50 نسبة مستهدفة احتمالية وليست ضمانًا لتطابق العددين تمامًا.

**نافذة النتيجة:** 7 × 24 ساعة كاملة بعد `t0`، بحد بداية مشمول وحد نهاية غير مشمول، وبطوابع زمنية موحدة UTC. نعد أي طلب مدفوع ناجح للحساب نفسه في الموقع أو التطبيق، سواء عاد عبر الرسالة أو مباشرة، وحتى إذا عدّل محتويات السلة أو أنشأ سلة جديدة. لذلك يقيس المشروع عودة المستخدم للشراء بعد ترك السلة، ولا يدعي استرجاع السلة الأصلية بعناصرها نفسها. نزيل تكرار أحداث الدفع باستخدام `order_id`. لا تُعد محاولة دفع فاشلة أو طلب غير مدفوع شراءً. يبقى `ordered=1` إذا أُلغي الطلب لاحقًا؛ الإلغاء مقياس حماية منفصل. لا نقيس النتيجة من وقت فتح الرسالة، وننتظر اكتمال نافذة آخر مستخدم قبل التحليل في المرحلة اللاحقة. يوضح القسم 3 مدة التسجيل وعدد المستخدمين.

**جميع المستخدمين في المقارنة (Intention-to-treat, ITT):** مقام نسبة الشراء في كل مجموعة هو كل المستخدمين المؤهلين الموزعين إليها. لا نحذف من لم تُسلّم رسالته أو لم يفتح أو ينقر، ولا من ألغى الاشتراك أو اشترى قبل تنفيذ الإرسال. في البيانات المكتملة يكون عدم وجود طلب مدفوع خلال النافذة صفرًا، أما تعطل القياس أو فقدان السجلات فليس دليلًا على عدم الشراء ويجب فحصه قبل الاستنتاج لاحقًا. الفتح والنقر معلومات عن التنفيذ داخل B فقط وليسا شرطًا للدخول في التحليل.

**تعريف الحماية دون انتقاء لاحق:** الربح هو هامش أول طلب مدفوع خلال النافذة ناقص تكاليفه وتكلفة محاولة البريد في B، مع صفر لإيراد غير المشتري وبقاء تكلفة البريد عليه إن وُجدت. يوثق القسم 5 هامش الربح وتكلفة المحاولة المفترضين في المحاكاة. يقاس الدعم والإلغاء على مستوى كل مستخدم موزع وفي نافذة الأيام السبعة نفسها. الإلغاء هنا إلغاء أول طلب مدفوع داخل النافذة قبل نهايتها؛ الإلغاءات اللاحقة وأثر الولاء طويل المدى خارج نطاق القياس الحالي. متوسط قيمة الطلب المشروط بالشراء وصفي لأن التذكير قد يغير هوية المشترين.

````python
import warnings
warnings.filterwarnings("ignore")

import html
from IPython.display import HTML, display

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from cycler import cycler
from matplotlib import ft2font
from scipy import stats

# الإصدارات الحديثة من matplotlib تعرض العربية مباشرة، والقديمة تحتاج مكتبتين صغيرتين
NATIVE_ARABIC = hasattr(ft2font, "__libraqm_version__")
if not NATIVE_ARABIC:
    try:
        import arabic_reshaper
        from bidi.algorithm import get_display
    except ImportError:
        import subprocess
        import sys
        subprocess.run([sys.executable, "-m", "pip", "install", "-q", "arabic-reshaper", "python-bidi"], check=True)
        import arabic_reshaper
        from bidi.algorithm import get_display

rng = np.random.default_rng(2026)

# ألوان البرنامج: A تركوازي، وB برتقالي، والقيمة الحقيقية كحلي، والطريقة الساذجة رمادي
TEAL, ORANGE, NAVY, GRAY = "#0E7490", "#E58A1F", "#0F2A3F", "#A9B4BC"
plt.rcParams.update({
    "figure.figsize": (9, 4.5), "figure.dpi": 110,
    "axes.prop_cycle": cycler(color=[TEAL, ORANGE, NAVY, GRAY]),
    "axes.edgecolor": "#56656F", "axes.labelcolor": NAVY, "text.color": NAVY,
    "xtick.color": "#3D4F5C", "ytick.color": "#3D4F5C",
    "axes.grid": True, "grid.color": "#D8E1E7", "grid.linewidth": 0.8,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.titlesize": 13, "axes.titleweight": "bold", "font.size": 11,
    "legend.frameon": False,
})
pd.set_option("display.precision", 4)


def ar(text):
    """يجهز النص العربي ليظهر بالشكل والاتجاه الصحيحين داخل الرسوم."""
    text = str(text)
    return text if NATIVE_ARABIC else get_display(arabic_reshaper.reshape(text))


def say(*parts, sep=" "):
    """يطبع نصًا عربيًا من اليمين إلى اليسار."""
    text = html.escape(sep.join(str(p) for p in parts))
    display(HTML(f'<div dir="rtl" style="white-space: pre-wrap; line-height: 1.7;">{text}</div>'))

import statsmodels.formula.api as smf
from statsmodels.stats.multitest import multipletests
from IPython.display import Markdown, display


def n_per_group_proportions(p_control, p_treatment, alpha=0.05, power=0.80):
    z_a, z_b = stats.norm.ppf(1 - alpha / 2), stats.norm.ppf(power)
    var = p_control * (1 - p_control) + p_treatment * (1 - p_treatment)
    return int(np.ceil((z_a + z_b) ** 2 * var / (p_treatment - p_control) ** 2))


def mde_proportions(p_control, n_per_group, alpha=0.05, power=0.80):
    z_a, z_b = stats.norm.ppf(1 - alpha / 2), stats.norm.ppf(power)
    return (z_a + z_b) * np.sqrt(2 * p_control * (1 - p_control) / n_per_group)


def duration_days(n_per_group, daily_new_users, groups=2, round_to_weeks=True):
    days = int(np.ceil(n_per_group * groups / daily_new_users))
    return int(np.ceil(days / 7) * 7) if round_to_weeks else days


def srm_check(n_a, n_b, expected_share_b=0.5, threshold=0.001):
    total = n_a + n_b
    chi2, p = stats.chisquare([n_a, n_b], f_exp=[total * (1 - expected_share_b), total * expected_share_b])
    verdict = "خلل في التوزيع (SRM)، لا تثق بالنتائج" if p < threshold else "التوزيع سليم"
    say(f"A={n_a:,}   B={n_b:,}   χ²={chi2:.2f}   p={p:.2g}   النتيجة: {verdict}")
    return p


def diff_ci(a, b, alpha=0.05):
    """الفرق بين B وA مع فترة الثقة وقيمة p (بالتقريب الطبيعي)."""
    a, b = np.asarray(a, dtype=float), np.asarray(b, dtype=float)
    diff = b.mean() - a.mean()
    se = np.sqrt(a.var(ddof=1) / len(a) + b.var(ddof=1) / len(b))
    z = stats.norm.ppf(1 - alpha / 2)
    return {"الفرق": diff, "أدنى 95%": diff - z * se, "أعلى 95%": diff + z * se, "SE": se,
            "p": 2 * stats.norm.sf(abs(diff / se))}


def decide(ci_low, ci_high, mde, guardrails_ok=True):
    if ci_high < 0:
        return "لا نطلق، لأن النسخة الجديدة أضرت بالمقياس الرئيسي."
    if ci_low > 0 and not guardrails_ok:
        return "لا نطلق الآن، لأن المقياس الرئيسي تحسن لكن أحد مقاييس الحماية ساء. نعالج المشكلة ثم نعيد التحقق."
    if ci_low > 0:
        return "نطلق، لأن الأثر إيجابي بثقة ومقاييس الحماية سليمة."
    if ci_high < mde:
        return "لا نطلق، لأننا استبعدنا وجود أثر بالحجم الذي يهمنا."
    return "لا نقرر بعد. نمدد التجربة، أو نعيد تصميمها بقوة أعلى."


say("الأدوات جاهزة")
````

### الخطة المثبتة قبل توليد البيانات

1. **سلامة البيانات والتوزيع (SRM).** نطابق صفوف التحليل مع سجل التوزيع وأفواج التسجيل، ونتحقق من تفرد المعرفات، وثبات النسخة، واكتمال متابعة 7 أيام، واتساق النتائج والتكاليف. نقارن العددين بتخصيص 50/50 باختبار كاي تربيع عند `p < 0.001` حسب القالب. إذا أخفق فحص السلامة أو التوزيع، نوقف حساب الأثر ونحقق في السبب. نتحقق أيضًا من بلوغ 14,749 لكل مجموعة.
2. **المقياس الرئيسي.** تحليل حسب التخصيص (ITT) لجميع المؤهلين. نقدر فرق احتمال طلب مدفوع واحد على الأقل خلال 7 أيام، `B − A` بالنقاط المئوية، بفترة ثقة 95% واختبار ثنائي الطرف عند 5% بتقريب القالب. نستخدم التقدير الخام في قاعدة القرار. نحسب CUPED باستعمال عدد طلبات المستخدم في الـ90 يومًا السابقة للتوزيع كتقدير داعم، ونقارن الخطأ المعياري قبل الضبط وبعده.
3. **الثانوية.** نقارن إيراد أول طلب مدفوع لكل مستخدم (صفر لغير المشتري) ومتوسط قيمة أول طلب بين المشترين. نعرض فرق المتوسطات والفترات الاسمية 95% وقيم `p` المعدلة بطريقة Benjamini–Hochberg للمقارنتين عند FDR = 5%. متوسط قيمة الطلب بين المشترين مقارنة وصفية مشروطة؛ قد يختلف تكوين المشترين بين النسختين.
4. **الحماية.** حدود الضرر العملية التالية افتراضات تعليمية مسبقة. جميع المقامات تشمل الموزعين. نعتمد الحد الأسوأ من فترة الثقة 95% لنقرر هل استبعدنا ضررًا يتجاوز الهامش، والحد المقابل لمعرفة هل تدعم الفترة وقوعه. إذا قطعت الفترة هامش الضرر فالحماية غير حاسمة.

   | المقياس | هامش الضرر المقبول لـB − A |
   |---|---:|
   | الربح لكل مستخدم، بعد تكلفة محاولة البريد | انخفاض حتى 0.10 ريال لكل مستخدم |
   | إلغاء أول طلب خلال النافذة، لكل مستخدم | ارتفاع حتى 0.30 نقطة مئوية |
   | التواصل مع الدعم خلال النافذة، لكل مستخدم | ارتفاع حتى 0.50 نقطة مئوية |

   تُفحص الشكاوى وإلغاء الاشتراك من سجلات البريد في تجربة فعلية. محاكاة اليوم الثالث تغطي مقاييس الحماية الثلاثة في الجدول.
5. **توقيت القراءة والحداثة.** قراءة واحدة بعد 21 يوم تسجيل وإكمال آخر نافذة 7 أيام. نعرض الأثر حسب يوم التوزيع ونقارن أول 3 أيام ببقية الأيام كتشخيص استكشافي للحداثة، مع فترة ثقة 95% لفرق الأثر بين الفترتين. يحكم القرار الأثر الكلي المقرر مسبقًا. بعد التحليل نقارن التقدير بمتوسط الأثر الحقيقي الذي أدخلناه في المحاكاة.
6. **قاعدة القرار للتقرير.** تتوقف القراءة عند فشل فحوص الثقة. إذا كان حد الثقة الأعلى للمقياس الرئيسي أقل من الصفر أو دلّ أحد مقاييس الحماية على ضرر يتجاوز هامشه، فلا ندعم التوسع. إذا كان الحد الأعلى للأثر أقل من نقطة مئوية، استبعدنا الفائدة المستهدفة. يدعم التحليل الانتقال إلى اختبار ميداني عندما يكون الحد الأدنى للأثر فوق الصفر والتقدير الخام نقطة مئوية أو أكثر وتستبعد كل الحمايات تجاوز حدود الضرر. بخلاف ذلك تبقى الأدلة غير حاسمة للتوسع. توصية المشروع تُكتب في اليوم الرابع.

## 5. إنشاء البيانات بالمحاكاة

ندخل 1,500 مستخدم اصطناعي مؤهل يوميًا لمدة 21 يومًا، بإجمالي 31,500. نعطي كل مستخدم معرفًا فريدًا ووقت توزيع، ونثبت نسخته بدالة SHA-256 حسب تصميم القسم 2. يمثل كل صف نافذة متابعة كاملة مدتها 7 أيام من وقت التوزيع. نولد الشراء والدعم والإلغاء كمؤشرات لهذه النافذة، من دون توليد أحداث سلة أو سجلات رسائل تفصيلية.

نفترض في `CONFIG` أثرًا مستقرًا قدره **0.80 نقطة مئوية** للتذكير، وزيادة حداثة مؤقتة ترفع الأثر المفترض إلى **1.20 نقطة** في اليوم الأول ثم تتناقص أسيًا. النشاط الكامن يؤثر في احتمال الشراء وفي الطلبات السابقة، ما يعطي CUPED متغيرًا سابقًا للتوزيع. نحفظ متوسط الأثر الحقيقي المطبق على المجتمع الاصطناعي للمقارنة بعد التحليل.

لكل شراء قيمة أول طلب مدفوع (بمتوسط مدخل 120 ريالًا). نسجل إيراد الطلب المدفوع حتى لو أُلغي لاحقًا، ونفترض مساهمة صفرية للطلب الملغى في الربح. هامش الطلب المكتمل 25%. نفترض محاولة بريد واحدة لكل مستخدم في B بتكلفة 0.02 ريال، بما في ذلك من لم يشترِ. نفترض احتمال إلغاء 5% بين المشترين في النسختين، واحتمال دعم 3% في A و3.1% في B. الإلغاء والدعم في التحليل محسوبان لكل مستخدم موزع.

بذرة المحاكاة **2026**. الأثر والتكاليف والحداثة افتراضات تعليمية. تُعرض آثار المقارنة بعد فحوص الثقة.

</div>

````python
from hashlib import sha256


def assignment(user_id, experiment_id, treatment_share):
    key = f"{experiment_id}:{user_id}".encode("utf-8")
    bucket = int.from_bytes(sha256(key).digest()[:8], "big")
    return "B" if bucket < treatment_share * (2 ** 64) else "A"


def make_roster(cfg, registration_days):
    daily_users = cfg["daily_new_users"]
    n = registration_days * daily_users
    user_ids = [f"sim-{i:06d}" for i in range(1, n + 1)]
    cohort_day = np.repeat(np.arange(1, registration_days + 1), daily_users)
    fraction_of_day = np.tile((np.arange(daily_users) + 0.5) / daily_users, registration_days)
    return pd.DataFrame({
        "user_id": user_ids,
        "day": cohort_day,
        "t0_day": cohort_day - 1 + fraction_of_day,
        "variant": [assignment(uid, cfg["experiment_id"], cfg["treatment_share"]) for uid in user_ids],
    })


def simulate_experiment(cfg, roster):
    random = np.random.default_rng(cfg["seed"])
    n = len(roster)
    treated = roster["variant"].eq("B").to_numpy(dtype=int)
    activity = random.normal(0, 1, n)
    base = cfg["baseline_conversion"]
    logit = np.log(base / (1 - base)) + cfg["activity_strength"] * activity
    p_a = 1 / (1 + np.exp(-logit))
    p_a = np.clip(p_a * base / p_a.mean(), 0, 1)
    pre_orders = random.poisson(5 * p_a / base)
    lift = cfg["true_lift_abs"] * (1 + cfg["novelty_boost"] * np.exp(-(roster["day"].to_numpy() - 1) / 3))
    p_b = np.clip(p_a + lift, 0, 1)
    ordered = random.binomial(1, np.where(treated == 1, p_b, p_a))
    value = np.maximum(10, random.normal(
        cfg["order_value_mean"] + cfg["order_value_effect"] * treated, cfg["order_value_sd"], n,
    ))
    revenue = ordered * value
    cancelled = ordered * random.binomial(1, cfg["cancel_rate"] + cfg["cancel_effect"] * treated)
    support = random.binomial(1, cfg["support_rate"] + cfg["support_effect"] * treated)
    profit = revenue * cfg["margin"] * (1 - cancelled) - treated * cfg["email_cost_per_attempt"]

    data = roster.assign(
        followup_end_day=roster["t0_day"] + cfg["observation_window_days"],
        pre_orders_90d=pre_orders,
        ordered=ordered,
        order_value=np.where(ordered == 1, value, np.nan),
        revenue=revenue,
        cancelled=cancelled,
        contacted_support=support,
        email_attempts=treated,
        profit=profit,
    )
    return data, float(np.mean(p_b - p_a))


trust_passed = False
roster = make_roster(CONFIG, days)
data, simulation_truth = simulate_experiment(CONFIG, roster)
say(f"تولّد {len(data):,} صفًا اصطناعيًا على مدى {days} يوم تسجيل ببذرة {CONFIG['seed']}.")
say(f"بيئة المحاكاة تستخدم NumPy {np.__version__} وpandas {pd.__version__}.")
````

## 6. فحوص الثقة

نطابق البيانات مع سجل التوزيع قبل تحليل الأثر. اختبار SRM يتحقق من اتساق الأعداد مع التوزيع 50/50. فحص السجل مستقل عنه، فقد يفقد الملف مستخدمين من المجموعتين بنسب متشابهة. أي إخفاق يوقف التنفيذ قبل القسم 7.

[مرجع تشخيص SRM، Microsoft Research](https://www.microsoft.com/en-us/research/articles/diagnosing-sample-ratio-mismatch-in-a-b-testing/)

</div>

````python
def check_experiment(data, roster, cfg, registration_days, minimum_per_group):
    counts = data["variant"].value_counts().reindex(["A", "B"], fill_value=0)
    if counts.min() == 0:
        raise ValueError("يجب أن تحتوي البيانات على المجموعتين A وB.")

    roster_fields = ["user_id", "day", "t0_day", "variant"]
    observed = data[roster_fields].sort_values("user_id").reset_index(drop=True)
    assigned = roster[roster_fields].sort_values("user_id").reset_index(drop=True)
    daily = data.groupby("day").size().reindex(range(1, registration_days + 1), fill_value=0)
    binary = ["ordered", "cancelled", "contacted_support", "email_attempts"]
    bought = data["ordered"].eq(1)
    expected_profit = data["revenue"] * cfg["margin"] * (1 - data["cancelled"]) - data["email_attempts"] * cfg["email_cost_per_attempt"]

    checks = {
        "صف فريد لكل مستخدم ومطابقة سجل التوزيع": data["user_id"].is_unique and observed.equals(assigned),
        "21 فوجًا كاملًا وتدفق التسجيل المخطط": len(data) == registration_days * cfg["daily_new_users"] and daily.eq(cfg["daily_new_users"]).all(),
        "اكتمال سبعة أيام من التوزيع لكل مستخدم": np.allclose(data["followup_end_day"] - data["t0_day"], cfg["observation_window_days"]) and data["followup_end_day"].le(registration_days + cfg["observation_window_days"]).all(),
        "بلوغ الحد الأدنى للعينة في المجموعتين": counts.min() >= minimum_per_group,
        "مؤشرات ثنائية كاملة وطلبات سابقة صالحة": data[binary].isin([0, 1]).all().all() and data["pre_orders_90d"].ge(0).all(),
        "إلغاء صفر لغير المشترين وقيمة طلب متسقة": data["cancelled"].le(data["ordered"]).all() and data.loc[bought, "order_value"].gt(0).all() and data.loc[~bought, "order_value"].isna().all(),
        "إيراد وربح ومحاولة بريد متسقة": np.allclose(data["revenue"], data["order_value"].fillna(0)) and np.allclose(data["profit"], expected_profit) and data["email_attempts"].eq(data["variant"].eq("B").astype(int)).all(),
    }
    srm_p = srm_check(int(counts["A"]), int(counts["B"]), cfg["treatment_share"], cfg["srm_alpha"])
    checks["توزيع 50/50 وفق اختبار SRM"] = srm_p >= cfg["srm_alpha"]
    display(pd.Series(checks, name="نجح").to_frame())
    if not all(checks.values()):
        failed = "، ".join(label for label, passed in checks.items() if not passed)
        raise ValueError(f"توقف التحليل بسبب إخفاق {failed}.")
    return counts, srm_p


trust_passed = False
counts, srm_p = check_experiment(data, roster, CONFIG, days, n_per_group)
trust_passed = True
say(f"حصة B الفعلية {counts['B'] / len(data):.2%}. اجتازت البيانات فحوص الثقة.")
````

## 7. التحليل

### 7.1 المقياس الرئيسي، مع CUPED

</div>

````python
if not trust_passed:
    raise RuntimeError("شغّل فحوص الثقة واجتزها قبل التحليل.")

A = data.loc[data["variant"].eq("A")]
B = data.loc[data["variant"].eq("B")]
alpha = CONFIG["alpha"]
primary = diff_ci(A["ordered"], B["ordered"], alpha)

pre = data["pre_orders_90d"]
pre_variance = pre.var(ddof=1)
theta = np.cov(data["ordered"], pre)[0, 1] / pre_variance if pre_variance > 0 else 0.0
adjusted = data["ordered"] - theta * (pre - pre.mean())
primary_cuped = diff_ci(adjusted.loc[A.index], adjusted.loc[B.index], alpha)
variance_reduction = 1 - (primary_cuped["SE"] / primary["SE"]) ** 2

say(f"أتم الشراء {A['ordered'].sum():,} من {len(A):,} في A ({A['ordered'].mean():.2%})، و{B['ordered'].sum():,} من {len(B):,} في B ({B['ordered'].mean():.2%}).")
say(f"انخفاض تباين مقدّر الأثر بعد CUPED بلغ {variance_reduction:.2%}، من مقارنة مربعي الخطأ المعياري.")
primary_table = pd.DataFrame([primary, primary_cuped], index=["خام (ITT)", "مع CUPED"])
primary_table[["الفرق", "أدنى 95%", "أعلى 95%", "SE"]] *= 100
say("الفرق وحدود الثقة والخطأ المعياري في الجدول بالنقاط المئوية.")
display(primary_table.round(4))
````

### 7.2 المقاييس الثانوية، مع تصحيح الاختبارات المتعددة

</div>

````python
secondary = {
    "الإيراد لكل مستخدم (ريال)": diff_ci(A["revenue"], B["revenue"], alpha),
    "متوسط أول طلب بين المشترين (ريال)": diff_ci(A["order_value"].dropna(), B["order_value"].dropna(), alpha),
}
sec = pd.DataFrame(secondary).T
rejected, adjusted_p, _, _ = multipletests(sec["p"], alpha=alpha, method="fdr_bh")
sec["p بعد BH"] = adjusted_p
sec["دال بعد BH"] = rejected
sec["A"] = [A["revenue"].mean(), A["order_value"].mean()]
sec["B"] = [B["revenue"].mean(), B["order_value"].mean()]
display(sec[["A", "B", "الفرق", "أدنى 95%", "أعلى 95%", "p", "p بعد BH", "دال بعد BH"]].round(4))
say("المقارنة الثانية وصفية بين المشترين. فترات الثقة اسمية، ويطبق BH على قيم p في المقارنتين.")
````

### 7.3 مقاييس الحماية

</div>

````python
guardrails = {
    "profit": ("الربح لكل مستخدم", -CONFIG["profit_loss_margin"], 1, "ريال لكل مستخدم"),
    "cancelled": ("خطر إلغاء أول طلب لكل مستخدم", CONFIG["cancel_increase_margin"], 100, "نقطة مئوية"),
    "contacted_support": ("التواصل مع الدعم لكل مستخدم", CONFIG["support_increase_margin"], 100, "نقطة مئوية"),
}
rows, guard_status = [], {}
for col, (label, margin, scale, unit) in guardrails.items():
    result = diff_ci(A[col], B[col], alpha)
    low, high = result["أدنى 95%"], result["أعلى 95%"]
    safe = low >= margin if margin < 0 else high <= margin
    harmful = high < margin if margin < 0 else low > margin
    status = "سليم" if safe else "ضرر يتجاوز الهامش" if harmful else "غير حاسم"
    guard_status[label] = status
    rows.append({"المقياس": label, "A": A[col].mean(), "B": B[col].mean(), **result,
                 "الهامش": margin, "الوحدة": unit, "المضاعف للعرض": scale, "الحالة": status})

guard_table = pd.DataFrame(rows).set_index("المقياس")
display_cols = ["A", "B", "الفرق", "أدنى 95%", "أعلى 95%", "الهامش"]
guard_display = guard_table[display_cols].mul(guard_table["المضاعف للعرض"], axis=0)
guard_display["وحدة الفرق"] = guard_table["الوحدة"]
guard_display["الحالة"] = guard_table["الحالة"]
display(guard_display.round(4))
say("مستويات الإلغاء والدعم في A وB معروضة بالنسبة المئوية، وفروقها بالنقاط المئوية. جميع المقامات تشمل الموزعين.")
````

````python
daily_rows = []
for day, cohort in data.groupby("day"):
    result = diff_ci(cohort.loc[cohort["variant"].eq("A"), "ordered"],
                     cohort.loc[cohort["variant"].eq("B"), "ordered"], alpha)
    daily_rows.append({"day": day, **result})
daily_effect = pd.DataFrame(daily_rows).set_index("day")

fig, ax = plt.subplots(figsize=(8.6, 4.2), dpi=100)
ax.errorbar(daily_effect.index, daily_effect["الفرق"] * 100,
            yerr=(daily_effect["أعلى 95%"] - daily_effect["الفرق"]) * 100,
            fmt="o-", markersize=3.5, linewidth=1, elinewidth=0.9, capsize=2,
            color=TEAL, label="Cohort estimate and 95% CI")
ax.axhline(primary["الفرق"] * 100, color=ORANGE, linestyle="--", label="Overall ITT")
ax.axhline(0, color=GRAY, linewidth=0.8)
ax.set(xlabel="Randomization day", ylabel="Purchase effect (percentage points)",
       title="Purchase effect by registration cohort", xticks=[1, 3, 7, 14, 21])
ax.legend(loc="best")
fig.tight_layout()
display(fig)
plt.close(fig)

early = diff_ci(A.loc[A["day"].le(3), "ordered"], B.loc[B["day"].le(3), "ordered"], alpha)
late = diff_ci(A.loc[A["day"].gt(3), "ordered"], B.loc[B["day"].gt(3), "ordered"], alpha)
novelty_delta = early["الفرق"] - late["الفرق"]
novelty_half_width = stats.norm.ppf(1 - alpha / 2) * np.hypot(early["SE"], late["SE"])
novelty_low = novelty_delta - novelty_half_width
novelty_high = novelty_delta + novelty_half_width
say(f"الأثر في الأيام 1–3 بلغ {early['الفرق'] * 100:+.2f} نقطة، وفي الأيام 4–21 بلغ {late['الفرق'] * 100:+.2f} نقطة.")
say(f"فرق الأثر بين الفترتين {novelty_delta * 100:+.2f} نقطة (95% من {novelty_low * 100:+.2f} إلى {novelty_high * 100:+.2f}).")
if novelty_low <= 0 <= novelty_high:
    say("فترة الثقة تشمل الصفر، لذلك فرق الفترتين غير حاسم في هذه العينة.")

primary_low, primary_high = primary["أدنى 95%"], primary["أعلى 95%"]
truth_covered = primary_low <= simulation_truth <= primary_high
coverage_text = "يقع داخل" if truth_covered else "يقع خارج"
say(f"متوسط الأثر الحقيقي المطبق في المحاكاة {simulation_truth * 100:+.3f} نقطة مئوية، و{coverage_text} فترة الثقة الخام.")

guardrails_ok = all(status == "سليم" for status in guard_status.values())
if primary_high < 0 or any(status == "ضرر يتجاوز الهامش" for status in guard_status.values()):
    day3_decision = "تدعم النتائج المحاكية وجود ضرر يوقف التوسع."
elif primary_high < CONFIG["mde_abs"]:
    day3_decision = "استبعدت الفترة فائدة تبلغ الحد العملي المستهدف."
elif primary_low > 0 and primary["الفرق"] >= CONFIG["mde_abs"] and guardrails_ok:
    day3_decision = "حقق السيناريو المحاكى شروط الانتقال إلى اختبار ميداني."
else:
    day3_decision = "الأدلة غير حاسمة وفق شروط التوسع المثبتة مسبقًا."
say(day3_decision)
````

## 8. (اختياري) إذا لم تستطع إجراء تجربة: الفروق في الفروق

إذا كان مشروعك سياسة طُبقت في مناطق دون غيرها، فعدّل الإعدادات في الخلية التالية، وطبق خطوات اليوم الثالث نفسها.

</div>

````python
def simulate_policy_panel(n_units=20, n_treated=6, n_periods=24, policy_period=13, effect=8.0, seed=32):
    r = np.random.default_rng(seed)
    level = r.normal(100, 15, n_units)
    shock = r.normal(0, 2, n_periods)
    rows = []
    for i in range(n_units):
        for t in range(1, n_periods + 1):
            treated, post = int(i < n_treated), int(t >= policy_period)
            y = level[i] + 0.8 * t + shock[t - 1] + effect * treated * post + r.normal(0, 3)
            rows.append({"unit": i, "period": t, "treated": treated, "post": post, "y": y})
    df = pd.DataFrame(rows)
    df["treated_post"] = df["treated"] * df["post"]
    return df


policy = simulate_policy_panel()
twfe = smf.ols("y ~ treated_post + C(unit) + C(period)", data=policy).fit(
    cov_type="cluster", cov_kwds={"groups": policy["unit"]})
lo, hi = twfe.conf_int().loc["treated_post"]
say(f"أثر السياسة: {twfe.params['treated_post']:.2f}، وفترة الثقة 95% من {lo:.2f} إلى {hi:.2f}")
````
